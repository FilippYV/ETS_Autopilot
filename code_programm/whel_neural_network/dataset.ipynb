```python
import numpy as np
import torch
from ultralytics import YOLO
from code_programm.path import get_path_weight_model
import pandas as pd
import os
import torch
import cv2
from torchvision.datasets import ImageFolder
import torchvision.transforms as transform
```

```python
wheel_ets_train_x = pd.DataFrame(columns=[i for i in range(128*128)])
```

```python
path = [r'D:\Dataset_for_autopilot\2024-03-29 07-01-32',
        r'D:\Dataset_for_autopilot\2024-03-29 06-57-45']
```

```python
os.path.join(r'D:\Dataset_for_autopilot\2024-03-29 07-04-10', f'road')
```

```python
model = YOLO(get_path_weight_model('best_v2_n.pt'))
for i in path:
    path_i = os.path.join(i, f'road')
    if os.path.exists(f'{path_i}') and os.path.isdir(f'{path_i}'):
        png_files = [os.path.join(path_i, file) for file in os.listdir(path_i) if file.endswith('.png')]
        print("Полные пути к файлам в папке:")
        for file_path in png_files:
            print(file_path)
    else:
        print("Указанный путь не существует или не является папкой.")
    
    size = 512
    combined_mask = np.zeros((size, size))    
    for file in png_files:
        results = model(file, imgsz=512, conf=0.65, show=False, device='cuda')
    
        combined_mask -= combined_mask * 0.5
    
        if results[0].masks is not None:
            for i in results[0].masks.data:
                combined_mask += i.cpu().numpy()
        new_image = cv2.resize(combined_mask, (128, 128))
        new_index = len(wheel_ets_train_x)
    
        # Добавьте новую строку в DataFrame с использованием метода loc
        name = file.split(f'\\')[-1].replace('.png', '')
        wheel_ets_train_x.loc[name] = new_image.flatten()
```
