```python
import time

from mss import mss
import numpy as np
from PIL import ImageGrab
import pyautogui
import mss.tools as mss_tools
import cv2

from code_programm.path import (get_path_config_road_area_size, get_path_config_speed_area_size)
```

```python
def get_road_area_size():
    with open(get_path_config_road_area_size(), 'r') as file:
        lines = file.readlines()
    return [int(lines[0]), int(lines[1]), int(lines[2]), int(lines[3])]


def get_speed_area_size():
    with open(get_path_config_speed_area_size(), 'r') as file:
        lines = file.readlines()
    return [int(lines[0]), int(lines[1]), int(lines[2]), int(lines[3])]


road_area = get_road_area_size()
speed_area = get_speed_area_size()
```

```python
def pyautogui_s():
    screenshot_pyautogui = pyautogui.screenshot()
    x1 = screenshot_pyautogui.crop((road_area[0], road_area[1], road_area[0] + road_area[2], road_area[1] + road_area[3]))
    x2 = screenshot_pyautogui.crop((speed_area[0], speed_area[1], speed_area[0] + speed_area[2], speed_area[1] + speed_area[3]))
    x1.save("pyautogui_s1.png")
    x2.save("pyautogui_s2.png")
```

```python
def imagegrab_s():
    screenshot_im_grab = ImageGrab.grab()
    road = screenshot_im_grab.crop((road_area[0], road_area[1], road_area[0] + road_area[2], road_area[1] + road_area[3]))
    speed = screenshot_im_grab.crop((speed_area[0], speed_area[1], speed_area[0] + speed_area[2], speed_area[1] + speed_area[3]))
    road.save(r"imagegrab_s1.png")
    speed.save(r"imagegrab_s2.png")
```

```python
start_time = time.time()
pyautogui_s()
print('Время:', time.time() - start_time)
print('pyautogui_s')
```

```python
start_time = time.time()
imagegrab_s()
print('Время:', time.time() - start_time)
print('pyautogui_s')
```

```python
array_1 = []
array_2 = []
for i in range(1000):
    start_time = time.time()
    pyautogui_s()
    array_1.append(time.time() - start_time)

    start_time = time.time()
    imagegrab_s()
    array_2.append(time.time() - start_time)
```

```python
print(sum(array_1) / len(array_1))
print(sum(array_2) / len(array_2))
```
