```python
import time
import dxcam
import numpy as np
from PIL import ImageGrab
import pyautogui
import cv2
from PIL import Image
import scipy.misc
import mss
import matplotlib.image as img


from code_programm.path import (get_path_config_road_area_size, get_path_config_speed_area_size)
```

```python
dxcam.device_info()
```

```python
def get_road_area_size():
    with open(get_path_config_road_area_size(), 'r') as file:
        lines = file.readlines()
    return [int(lines[1]), int(lines[1]) + int(lines[3]), int(lines[0]), int(lines[0]) + int(lines[2])]


def get_speed_area_size():
    with open(get_path_config_speed_area_size(), 'r') as file:
        lines = file.readlines()
    return [int(lines[1]), int(lines[1]) + int(lines[3]), int(lines[0]), int(lines[0]) + int(lines[2])]



road_area = get_road_area_size()
speed_area = get_speed_area_size()
```

```python
import os
def dell_files():
    os.remove('dxcam_s1.png')
    os.remove('dxcam_s2.png')
    os.remove('mss_s1.png')
    os.remove('mss_s2.png')
    os.remove('imagegrab_s1.png')
    os.remove('imagegrab_s2.png')
    os.remove('pyautogui_s1.png')
    os.remove('pyautogui_s2.png')
    
def dell_files_dxcam():
    os.remove('dxcam_s1.png')
    os.remove('dxcam_s2.png')
```

```python
def pyautogui_s():
    screenshot_pyautogui = pyautogui.screenshot()
    x1 = screenshot_pyautogui.crop((road_area[0], road_area[1], road_area[0] + road_area[2], road_area[1] + road_area[3]))
    x2 = screenshot_pyautogui.crop((speed_area[0], speed_area[1], speed_area[0] + speed_area[2], speed_area[1] + speed_area[3]))
    x1.save("pyautogui_s1.png")
    x2.save("pyautogui_s2.png")
```

```python
def imagegrab_s():
    screenshot_im_grab = ImageGrab.grab()
    road = screenshot_im_grab.crop((road_area[0], road_area[1], road_area[0] + road_area[2], road_area[1] + road_area[3]))
    speed = screenshot_im_grab.crop((speed_area[0], speed_area[1], speed_area[0] + speed_area[2], speed_area[1] + speed_area[3]))
    road.save(r"imagegrab_s1.png")
    speed.save(r"imagegrab_s2.png")
```

```python
def dxcam_s(camera):
    screenshot_dxcam = camera.get_latest_frame() 
    road =  screenshot_dxcam[road_area[1]:road_area[1] + road_area[3], road_area[0]:road_area[0] + road_area[2]]
    speed = screenshot_dxcam[speed_area[1]:speed_area[1] + speed_area[3], speed_area[0]:speed_area[0] + speed_area[2]]
    cv2.imshow('das', road)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
    cv2.imwrite('dxcam_s1.png', road)
    cv2.imwrite('dxcam_s2.png', speed)
```

```python
def mss_s():
    with mss.mss() as sct:
        road = np.array(sct.grab((road_area[0], road_area[1], road_area[0] + road_area[2], road_area[1] + road_area[3])))
        speed = np.array(sct.grab((speed_area[0], speed_area[1], speed_area[0] + speed_area[2], speed_area[1] + speed_area[3])))
        img.imsave('mss_s1.png', road)
        img.imsave('mss_s2.png', speed)
```

```python
start_time = time.time()
pyautogui_s()
print('Время:', time.time() - start_time)
print('pyautogui_s')
```

```python
start_time = time.time()
imagegrab_s()
print('Время:', time.time() - start_time)
print('pyautogui_s')
```

```python
camera = dxcam.create(device_idx=0, output_idx=0, output_color='BGRA')
camera.start(target_fps=222)
start_time = time.time()
dxcam_s(camera)
print('Время:', time.time() - start_time)
print('dxcam_s')
```

```python
start_time = time.time()
mss_s()
print('Время:', time.time() - start_time)
print('mss_s')
```

```python
dell_files()
array_1 = []
array_2 = []
array_3 = []
array_4 = []
for i in range(600):
    start_time = time.time()
    pyautogui_s()
    array_1.append(time.time() - start_time)

    start_time = time.time()
    imagegrab_s()
    array_2.append(time.time() - start_time)
    
    start_time = time.time()
    dxcam_s(camera)
    array_3.append(time.time() - start_time)
    
    start_time = time.time()
    mss_s()
    array_4.append(time.time() - start_time)
    dell_files()
```
