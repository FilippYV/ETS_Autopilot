```python
pairs_df = pd.DataFrame(columns=['value1', 'value2'])
wheel_ets_train_y = pd.DataFrame(columns=['value'])
```

```python
for path in paths:
    array = pd.DataFrame(columns=['value'])
    path_i = os.path.join(path, f'log.csv')
    with open(path_i, 'r') as file:
        for line in file:
            lines = line.strip().split(',')
            array.loc[lines[0]] = [lines[1]]
    df_pairs = pd.DataFrame({
    'value1': array['value'],
    'value2': array['value'].shift(-1)
    })
    
    # Удаляем последнюю строку, так как для нее пара не полная
    df_pairs = df_pairs.dropna()
    
    # Добавляем данные в основной DataFrame
    pairs_df = pd.concat([pairs_df, df_pairs], ignore_index=True)
    print(len(pairs_df))
    print(pairs_df.iloc[0])
    print(pairs_df.head())
    
    
    
```
